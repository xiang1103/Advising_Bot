### Try out langchain as the response generation to incorporate memory 

```python
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
import os
from dotenv import load_dotenv
from google import genai
load_dotenv()
gemini_key = os.getenv("Gemini_key")
```

```python
model = ChatGoogleGenerativeAI(
    api_key= gemini_key, 
    model="gemini-3-flash-preview",
    temperature=1.0,  # Gemini 3.0+ defaults to 1.0
    max_tokens=None,
    timeout=None,
    max_retries=2,
    # other params...
)
```

```python
messages = [
    (
        "system",
        "Describe the city of Paris to me. how does it relate to Ernest Hemmingway's 'A movable feast'",
        
    ),
    ("human")
]
ai_msg = model.invoke(messages)
```
